We'll read a DNA sequence and ask it a few simple questions — no setup required.


```python
# A DNA sequence is just text made of A, T, G and C
dna = "ATGCGTACGTTAGCCGTA"
print (dna)
print (len(dna))
```

```python
# Count each base
print ("A:", dna.count ("A"))
print ("t:", dna.count("T"))
print ("G:", dna.count("G"))
print("C:", dna.count("C"))
```

```python
# GC content — the percentage of G and C bases
gc = (dna.count("G") + dna.count("C")) / len(dna) * 100
print ("gc content:", round(gc,1), "%")
```
